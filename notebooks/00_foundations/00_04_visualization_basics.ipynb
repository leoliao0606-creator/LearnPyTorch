# 00.4 基础可视化（Visualization Basics）

这份 notebook 的目标是让你能把数据和训练结果“画出来”。  

不会画图，通常就很难判断：  

- 数据分布（data distribution）
- 类别是否不平衡（whether classes are imbalanced）
- 模型是否过拟合
- 预测误差是否有规律

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

## 1. 数据准备

为了专注于画图本身，这里直接构造几份小数据。  


```python
student_df = pd.DataFrame(
    {
        "hours": [1.2, 2.1, 2.8, 3.0, 1.7, 4.2, 3.6, 2.4, 1.4, 3.9],
        "attendance": [0.62, 0.75, 0.81, 0.88, 0.67, 0.96, 0.91, 0.79, 0.58, 0.93],
        "city": ["A", "B", "A", "B", "A", "C", "C", "B", "A", "C"],
        "passed": [0, 0, 1, 1, 0, 1, 1, 1, 0, 1],
    }
)

history_df = pd.DataFrame(
    {
        "epoch": np.arange(1, 11),
        "train_loss": [0.92, 0.76, 0.63, 0.54, 0.47, 0.41, 0.36, 0.31, 0.28, 0.24],
        "val_loss": [0.95, 0.81, 0.71, 0.65, 0.61, 0.60, 0.62, 0.66, 0.71, 0.78],
        "train_acc": [0.55, 0.62, 0.70, 0.76, 0.80, 0.84, 0.87, 0.90, 0.92, 0.94],
        "val_acc": [0.53, 0.60, 0.66, 0.70, 0.72, 0.73, 0.72, 0.71, 0.69, 0.67],
    }
)

cm = np.array([[4, 1], [2, 7]])

reg_df = pd.DataFrame(
    {
        "y_true": [2.0, 2.4, 3.1, 3.8, 4.2, 5.0, 5.5, 6.3],
        "y_pred": [2.1, 2.2, 3.0, 4.0, 4.1, 5.3, 5.2, 6.0],
    }
)

student_df
```

## 2. 分布图

数值型特征常先看分布。  

常见问题

- 值域大概在哪
- 是否偏斜（is it skewed?）
- 是否有异常值（are there possible outliers?）

```python
plt.figure(figsize=(7, 4))
sns.histplot(student_df["hours"], bins=6, kde=True)
plt.title("Study Hours Distribution / 学习时长分布")
plt.xlabel("hours")
plt.ylabel("count")
plt.show()
```

## 3. 类别计数图

当你想看类别是否平衡时，计数图非常直接。  


```python
plt.figure(figsize=(7, 4))
sns.countplot(data=student_df, x="city")
plt.title("Students per City / 各城市样本数")
plt.xlabel("city")
plt.ylabel("count")
plt.show()

print("城市频数 / city counts:")
print(student_df["city"].value_counts())
```

## 4. 训练曲线

overfitting 的核心工具之一。  

一个常见信号

- 训练损失继续下降（train loss keeps decreasing）
- 验证损失开始上升（validation loss starts increasing）

```python
plt.figure(figsize=(8, 4))
sns.lineplot(data=history_df, x="epoch", y="train_loss", marker="o", label="train_loss")
sns.lineplot(data=history_df, x="epoch", y="val_loss", marker="o", label="val_loss")
plt.title("Training vs Validation Loss / 训练与验证损失")
plt.xlabel("epoch")
plt.ylabel("loss")
plt.show()
```

## 5. 混淆矩阵

分类任务只看一个总准确率往往不够。  

confusion matrix 可以告诉你：

- 哪一类更容易被错分
- 错误主要发生在哪种方向

```python
plt.figure(figsize=(5, 4))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=[0, 1], yticklabels=[0, 1])
plt.title("Confusion Matrix / 混淆矩阵")
plt.xlabel("Predicted label / 预测标签")
plt.ylabel("True label / 真实标签")
plt.show()
```

## 6. 散点图

回归任务里，散点图常用来比较真实值  


```python
plt.figure(figsize=(5, 5))
sns.scatterplot(data=reg_df, x="y_true", y="y_pred", s=80)
line_min = min(reg_df.min())
line_max = max(reg_df.max())
plt.plot([line_min, line_max], [line_min, line_max], linestyle="--", color="red")
plt.title("True vs Predicted / 真实值与预测值")
plt.xlabel("y_true")
plt.ylabel("y_pred")
plt.show()
```

如果点越靠近对角线，通常说明预测越接近真实值。
