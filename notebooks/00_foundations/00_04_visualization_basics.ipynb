The goal of this notebook is to help you visualize both data and training results.

Without visualization, it is hard to judge:

- data distribution
- whether classes are imbalanced
- whether the model is overfitting
- whether prediction errors show a pattern

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

## Preparing Small Example Data

To focus on visualization itself, we construct a few small example datasets directly.


```python
student_df = pd.DataFrame(
    {
        "hours": [1.2, 2.1, 2.8, 3.0, 1.7, 4.2, 3.6, 2.4, 1.4, 3.9],
        "attendance": [0.62, 0.75, 0.81, 0.88, 0.67, 0.96, 0.91, 0.79, 0.58, 0.93],
        "city": ["A", "B", "A", "B", "A", "C", "C", "B", "A", "C"],
        "passed": [0, 0, 1, 1, 0, 1, 1, 1, 0, 1],
    }
)

history_df = pd.DataFrame(
    {
        "epoch": np.arange(1, 11),
        "train_loss": [0.92, 0.76, 0.63, 0.54, 0.47, 0.41, 0.36, 0.31, 0.28, 0.24],
        "val_loss": [0.95, 0.81, 0.71, 0.65, 0.61, 0.60, 0.62, 0.66, 0.71, 0.78],
        "train_acc": [0.55, 0.62, 0.70, 0.76, 0.80, 0.84, 0.87, 0.90, 0.92, 0.94],
        "val_acc": [0.53, 0.60, 0.66, 0.70, 0.72, 0.73, 0.72, 0.71, 0.69, 0.67],
    }
)

cm = np.array([[4, 1], [2, 7]])

reg_df = pd.DataFrame(
    {
        "y_true": [2.0, 2.4, 3.1, 3.8, 4.2, 5.0, 5.5, 6.3],
        "y_pred": [2.1, 2.2, 3.0, 4.0, 4.1, 5.3, 5.2, 6.0],
    }
)

student_df
```

## Distribution Plot

For numeric features, one of the first things to inspect is the distribution.

Common questions:

- what is the rough value range?
- is it skewed?
- are there possible outliers?

```python
plt.figure(figsize=(7, 4))
sns.histplot(student_df["hours"], bins=6, kde=True)
plt.title("Study Hours Distribution / study-hours distribution")
plt.xlabel("hours")
plt.ylabel("count")
plt.show()
```

## Count Plot

A count plot is a direct way to see whether categories are balanced.


```python
plt.figure(figsize=(7, 4))
sns.countplot(data=student_df, x="city")
plt.title("Students per City / samples per city")
plt.xlabel("city")
plt.ylabel("count")
plt.show()

print("city counts / city counts:")
print(student_df["city"].value_counts())
```

## Training Curves

Training curves are one of the core tools for detecting overfitting.

A common signal:

- train loss keeps decreasing
- validation loss starts increasing

```python
plt.figure(figsize=(8, 4))
sns.lineplot(data=history_df, x="epoch", y="train_loss", marker="o", label="train_loss")
sns.lineplot(data=history_df, x="epoch", y="val_loss", marker="o", label="val_loss")
plt.title("Training vs Validation Loss / training and validation loss")
plt.xlabel("epoch")
plt.ylabel("loss")
plt.show()
```

## Confusion Matrix

For classification, overall accuracy alone is often not enough.

confusion matrix / confusion matrix can tell you:

- which class is more often misclassified
- in which direction errors happen

```python
plt.figure(figsize=(5, 4))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=[0, 1], yticklabels=[0, 1])
plt.title("Confusion Matrix / confusion matrix")
plt.xlabel("Predicted label / predicted label")
plt.ylabel("True label / true label")
plt.show()
```

## Scatter Plot

In regression, scatter plots are often used to compare true values and predictions.


```python
plt.figure(figsize=(5, 5))
sns.scatterplot(data=reg_df, x="y_true", y="y_pred", s=80)
line_min = min(reg_df.min())
line_max = max(reg_df.max())
plt.plot([line_min, line_max], [line_min, line_max], linestyle="--", color="red")
plt.title("True vs Predicted / true vs predicted")
plt.xlabel("y_true")
plt.ylabel("y_pred")
plt.show()
```

The closer the points are to the diagonal line, the closer the predictions are to the true values.
