# 00.4 Visualization Basics

The goal of this notebook is to help you use plots as a modeling tool, not as decoration. Before training, plots help you understand feature ranges, class balance, and possible outliers. During training, plots help you see whether the model is improving or overfitting. After evaluation, plots such as confusion matrices help you understand which errors the model is making.

The habit to build is simple: when a number looks good or bad, make a plot that explains what kind of good or bad it is.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

## Preparing Small Example Data

To focus on visualization itself, we construct a few small example datasets directly.


```python
student_df = pd.DataFrame(
    {
        "hours": [1.2, 2.1, 2.8, 3.0, 1.7, 4.2, 3.6, 2.4, 1.4, 3.9],
        "attendance": [0.62, 0.75, 0.81, 0.88, 0.67, 0.96, 0.91, 0.79, 0.58, 0.93],
        "city": ["A", "B", "A", "B", "A", "C", "C", "B", "A", "C"],
        "passed": [0, 0, 1, 1, 0, 1, 1, 1, 0, 1],
    }
)

history_df = pd.DataFrame(
    {
        "epoch": np.arange(1, 11),
        "train_loss": [0.92, 0.76, 0.63, 0.54, 0.47, 0.41, 0.36, 0.31, 0.28, 0.24],
        "val_loss": [0.95, 0.81, 0.71, 0.65, 0.61, 0.60, 0.62, 0.66, 0.71, 0.78],
        "train_acc": [0.55, 0.62, 0.70, 0.76, 0.80, 0.84, 0.87, 0.90, 0.92, 0.94],
        "val_acc": [0.53, 0.60, 0.66, 0.70, 0.72, 0.73, 0.72, 0.71, 0.69, 0.67],
    }
)

cm = np.array([[4, 1], [2, 7]])

reg_df = pd.DataFrame(
    {
        "y_true": [2.0, 2.4, 3.1, 3.8, 4.2, 5.0, 5.5, 6.3],
        "y_pred": [2.1, 2.2, 3.0, 4.0, 4.1, 5.3, 5.2, 6.0],
    }
)

student_df
```

## Distribution Plot

For numeric features, a distribution plot is often the first visual check. It shows the rough value range, whether values are concentrated or spread out, whether the shape is skewed, and whether there are possible outliers. This matters because scale and outliers can affect preprocessing choices and model behavior.

```python
plt.figure(figsize=(7, 4))
sns.histplot(student_df["hours"], bins=6, kde=True)
plt.title("Study Hours Distribution / study-hours distribution")
plt.xlabel("hours")
plt.ylabel("count")
plt.show()
```

## Count Plot

A count plot is a direct way to see whether categories are balanced.


```python
plt.figure(figsize=(7, 4))
sns.countplot(data=student_df, x="city")
plt.title("Students per City / samples per city")
plt.xlabel("city")
plt.ylabel("count")
plt.show()

print("city counts / city counts:")
print(student_df["city"].value_counts())
```

## Training Curves

Training curves show how metrics change across epochs. If training loss keeps decreasing while validation loss stops improving or begins increasing, the model may be overfitting. If both training and validation metrics are poor, the model may be underfitting or the optimization setup may need attention.

```python
plt.figure(figsize=(8, 4))
sns.lineplot(data=history_df, x="epoch", y="train_loss", marker="o", label="train_loss")
sns.lineplot(data=history_df, x="epoch", y="val_loss", marker="o", label="val_loss")
plt.title("Training vs Validation Loss / training and validation loss")
plt.xlabel("epoch")
plt.ylabel("loss")
plt.show()
```

## Confusion Matrix

Overall accuracy tells you how often the model is correct, but it does not tell you which classes are confused with each other. A confusion matrix shows the direction of mistakes. In binary classification, it lets you see whether false positives or false negatives are the bigger problem. In multiclass classification, it can reveal specific class pairs that the model struggles to separate.

```python
plt.figure(figsize=(5, 4))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=[0, 1], yticklabels=[0, 1])
plt.title("Confusion Matrix / confusion matrix")
plt.xlabel("Predicted label / predicted label")
plt.ylabel("True label / true label")
plt.show()
```

## Scatter Plot

In regression, scatter plots are often used to compare true values and predictions.


```python
plt.figure(figsize=(5, 5))
sns.scatterplot(data=reg_df, x="y_true", y="y_pred", s=80)
line_min = min(reg_df.min())
line_max = max(reg_df.max())
plt.plot([line_min, line_max], [line_min, line_max], linestyle="--", color="red")
plt.title("True vs Predicted / true vs predicted")
plt.xlabel("y_true")
plt.ylabel("y_pred")
plt.show()
```

The closer the points are to the diagonal line, the closer the predictions are to the true values.

## Summary

The value of plots is not that they look nice, but that they help you diagnose problems. A distribution plot helps you inspect the range and shape of numeric values. A count plot helps you inspect category frequencies and class imbalance. Training curves help you detect overfitting when validation performance worsens while training performance keeps improving. A confusion matrix shows the direction of classification mistakes instead of hiding everything behind one accuracy number.
